# PENDAHULUAN
Kemajuan teknologi Artificial Intelligence (AI) telah membawa transformasi yang signifikan dalam berbagai disiplin ilmu, termasuk dalam tahapan eksplorasi dan pemodelan geofisika. Saat ini, AI sangat diandalkan untuk mengotomatisasi pengolahan data, mendeteksi pola anomali, serta mempercepat proses interpretasi kondisi bawah permukaan. Meskipun demikian, kualitas keluaran yang dihasilkan oleh AI sangat bergantung pada instruksi atau prompt yang diberikan oleh pengguna. Tanpa arahan yang terstruktur, AI berpotensi menghasilkan analisis yang bias, tidak akurat, atau tidak relevan secara fisis.

Untuk mengatasi permasalahan tersebut, tugas ini menyusun rancangan prompt menggunakan kerangka kerja (work specification) yang terdiri dari komponen Role (R), Task (T), Context (C), Input (I), Constraints (K), Output (O), Quality Criteria (Q), dan Verification (V). Pendekatan sistematis ini bertujuan untuk memastikan bahwa setiap perintah yang diproses oleh sistem AI memiliki batasan operasional yang terukur serta target capaian yang spesifik.

Penerapan kerangka kerja ini akan didemonstrasikan pada lima topik analisis geofisika, yaitu: identifikasi data Ground Penetrating Radar (GPR), pemodelan inversi data magnetik, pencitraan tomografi seismik, pemetaan potensi panas bumi (geotermal), serta analisis data mikroseismik untuk tinjauan kestabilan lahan. Melalui perancangan prompt ini, AI diharapkan dapat berfungsi sebagai asisten analitis yang presisi dalam membantu penyelesaian berbagai tantangan di bidang geosains.

# 1. ANALISIS DATA GROUND PENETRATING RADAR (GPR)
## Role
Kita berperan sebagai mahasiswa geofisika yang sedang menganalisis data gelombang elektromagnetik frekuensi tinggi untuk mendeteksi objek dangkal di bawah permukaan.

## Task
Mengidentifikasi keberadaan utilitas bawah tanah (seperti pipa atau kabel) serta batas lapisan dangkal dengan mengamati pola difraksi hiperbolik pada penampang radargram.

## Context
Metode Ground Penetrating Radar (GPR) memanfaatkan pancaran gelombang elektromagnetik untuk mencitrakan kondisi bawah permukaan secara dangkal. Ketika gelombang mengenai objek lokal seperti pipa atau rongga, sinyal tersebut akan dipantulkan kembali dan terekam dalam bentuk pola hiperbola pada radargram. Identifikasi pola ini sangat esensial untuk memetakan utilitas bawah tanah dan mitigasi sebelum proses penggalian dilakukan.

## Input
Data yang digunakan berupa profil radargram 2D yang menampilkan kedalaman (atau waktu tempuh pantulan) pada sumbu vertikal dan jarak lintasan pengukuran pada sumbu horizontal.

## Constraints
- Jangan langsung menyimpulkan jenis material (misalnya baja atau plastik) hanya dari bentuk hiperbola tanpa adanya data pendukung.
- Bedakan secara cermat antara pantulan objek target dengan derau (noise) atau pantulan dari permukaan tanah.
- Analisis harus didasarkan murni pada kejelasan kontras amplitudo yang terekam pada data.
- Jangan menyimpulkan keberadaan patahan geologi yang dalam jika data hanya mencitrakan beberapa meter pertama di bawah permukaan.
- Jika terdapat pola difraksi yang terpotong atau samar, nyatakan dengan jelas bahwa identifikasi objek tersebut belum dapat dipastikan.

## Output
Hasil analisis memuat:

- Posisi lintasan (jarak) ditemukannya pola hiperbola.
- Estimasi kedalaman target anomali.
- Indikasi batas antara lapisan tanah keras dan tanah penutup (topsoil).
- Penjelasan rasional mengenai alasan pola tersebut diidentifikasi sebagai anomali.
- Rekomendasi singkat terkait area yang aman untuk proses penggalian.

## Quality Criteria
Analisis harus disusun secara logis dan sesuai dengan prinsip perambatan gelombang elektromagnetik. Penentuan posisi objek harus selalu merujuk pada titik puncak (verteks) dari hiperbola. Gunakan bahasa yang faktual, hindari asumsi material yang tidak berdasar, dan selalu berikan batasan toleransi kesalahan pada data yang memiliki banyak derau.

## Verification
Lakukan pengecekan silang terhadap keseluruhan penampang radargram. Pastikan anomali hiperbolik yang diidentifikasi memiliki puncak yang tajam dengan lengan hiperbola yang simetris. Jika bentuknya asimetris, evaluasi kembali apakah pola tersebut berasal dari benda lokal atau sekadar perubahan kemiringan lapisan tanah.

# 2. PEMODELAN INVERSI DATA MAGNETIK
## Role
Kita berperan sebagai mahasiswa pemodelan geofisika yang bertugas memetakan sebaran batuan dengan tingkat kemagnetan tinggi di bawah permukaan.

## Task
Menganalisis peta anomali magnetik total untuk mengidentifikasi area prospek batuan beku intrusif atau potensi mineralisasi bijih besi.

## Context
Metode magnetik mengukur variasi medan magnet bumi yang diakibatkan oleh perbedaan nilai suseptibilitas batuan bawah permukaan. Batuan beku atau mineral feromagnetik umumnya menghasilkan anomali magnetik yang kuat berupa pasangan kutub positif dan negatif (dipol). Pemahaman terhadap pola dipol ini sangat membantu dalam melokalisasi posisi dan estimasi kedalaman benda penyebab anomali.

## Input
Data yang digunakan berupa peta kontur anomali magnetik total (Total Magnetic Intensity / TMI) yang menampilkan distribusi nilai medan magnet (dalam satuan nanoTesla/nT) beserta koordinat geografisnya.

## Constraints
- Jangan berasumsi bahwa area dengan anomali tertinggi otomatis merupakan pusat batuan beku tanpa melihat keberadaan anomali negatif sebagai pasangannya (kecuali jika survei dilakukan tepat di ekuator magnetik).
- Hindari interpretasi struktur tunggal apabila terdapat tumpang tindih (superposisi) dari beberapa anomali yang saling berdekatan.
- Pisahkan dengan jelas antara anomali regional (dalam dan meluas) dengan anomali residual (dangkal dan lokal).
- Analisis tidak boleh menyimpulkan volume pasti dari cadangan mineral, melainkan hanya memetakan potensi distribusinya.
- Jika pada peta teridentifikasi derau kultural (seperti jembatan baja atau jaringan rel kereta), kecualikan anomali tersebut dari interpretasi geologi.

## Output
Hasil analisis memuat:

- Lokasi pusat anomali dipol (pasangan positif-negatif).
- Perkiraan sebaran spasial batuan intrusi atau zona mineralisasi.
- Indikasi arah kelurusan (trend) struktur geologi berdasarkan pola kontur.
- Penjelasan singkat mengenai asumsi perbedaan nilai suseptibilitas antar batuan.
- Kesimpulan mengenai zona prospek yang paling direkomendasikan untuk survei lanjutan.

## Quality Criteria
Pemaparan analisis harus menyertakan prinsip dasar bentuk dipol magnetik dan memperhitungkan pengaruh letak lintang (latitude) terhadap bentuk anomali. Penjelasan harus diuraikan secara sistematis, mulai dari pengenalan anomali dominan hingga identifikasi tren struktur, dengan menggunakan terminologi geofisika yang tepat.

## Verification
Periksa kembali kesesuaian antara bentuk anomali pada peta dengan teori interpretasi magnetik. Pastikan tidak ada kekeliruan dalam membedakan efek anomali buatan manusia dengan target geologi yang sebenarnya. Jika pola anomali tampak tidak lazim, sarankan penerapan filter matematis (seperti reduksi ke kutub) untuk validasi data lebih lanjut.

# 3. PENCITRAAN TOMOGRAFI SEISMIK
## Role
Kita berperan sebagai mahasiswa peneliti bumi padat (solid earth) yang bertugas menafsirkan model kecepatan perambatan gelombang seismik dalam skala regional.

## Task
Menganalisis penampang kecepatan gelombang P dan S secara 3D untuk mengidentifikasi keberadaan zona subduksi lempeng atau letak kantong magma di bawah gunung api.

## Context
Tomografi seismik bekerja dengan prinsip yang menyerupai CT-Scan medis, namun menggunakan waktu tempuh gelombang gempa. Kecepatan gelombang seismik akan bervariasi bergantung pada suhu dan densitas medium yang dilaluinya. Area dengan anomali kecepatan tinggi umumnya merepresentasikan material padat dan dingin (seperti lempeng tektonik yang menunjam), sedangkan anomali kecepatan rendah sering diasosiasikan dengan material panas atau fluida (seperti dapur magma).

## Input
Data berupa penampang melintang (cross-section) atau penampang mendatar (depth-slice) yang menampilkan persentase distribusi anomali kecepatan gelombang primer (Vp) dan sekunder (Vs).

## Constraints
- Jangan menyebutkan angka kedalaman secara terlampau presisi jika resolusi grid model tomografi tergolong kasar.
- Jangan mendefinisikan zona kecepatan rendah mutlak sebagai "magma cair", melainkan pertimbangkan juga faktor keberadaan fluida hidrotermal.
- Interpretasi harus berpegang teguh pada skala warna anomali (positif atau negatif) yang tersaji pada model.
- Bedakan antara struktur tektonik utama dengan artefak pengolahan data matematis (smearing effect).
- Jika resolusi di suatu area sangat rendah akibat minimnya stasiun pencatat gempa, nyatakan secara eksplisit bahwa area tersebut memiliki tingkat reliabilitas yang rendah.

## Output
Hasil analisis memuat:

- Lokasi sebaran anomali kecepatan tinggi (zona padat/dingin) dan rendah (zona leleh/panas).
- Kemungkinan geometri lempeng subduksi yang menunjam ke dalam mantel bumi.
- Dugaan posisi akumulasi fluida atau kantong magma di wilayah vulkanik aktif.
- Korelasi antara pola anomali kecepatan di bawah permukaan dengan keberadaan jejeran gunung api di permukaan.
- Ringkasan kondisi tektonik regional berdasarkan model tomografi tersebut.

## Quality Criteria
Analisis harus logis dan mampu mengkorelasikan parameter fisika (kecepatan gelombang) dengan parameter geologi (suhu, densitas, fasa material). Laporan harus membedakan secara tegas mana zona pencitraan yang memiliki resolusi tinggi dan mana zona yang masih memiliki ketidakpastian tinggi.

## Verification
Bandingkan penampang anomali kecepatan dengan peta sebaran gunung api aktif atau distribusi titik hiposenter gempa bumi regional. Jika zona anomali kecepatan tinggi tidak berkorelasi dengan aktivitas seismik yang masuk akal, pertimbangkan kemungkinan bahwa area tersebut hanyalah artefak matematis dari proses inversi data.

# 4. PEMETAAN POTENSI PANAS BUMI (GEOTERMAL)
## Role
Kita berperan sebagai mahasiswa eksplorasi energi terbarukan yang sedang mengintegrasikan pengolahan awal data geofisika dan geokimia.

## Task
Mengevaluasi peta sebaran anomali suhu dan densitas kelurusan patahan untuk mendeliniasi area prospek reservoir panas bumi.

## Context
Sistem panas bumi pada umumnya ditandai oleh tiga komponen utama: sumber panas, lapisan batuan penudung (caprock dengan resistivitas rendah), dan reservoir permeabel yang sangat dikontrol oleh struktur patahan. Menggabungkan letak manifestasi di permukaan (seperti mata air panas atau fumarol) dengan peta kelurusan struktur dari data geofisika dapat memandu penentuan lokasi prospek pengeboran.

## Input
Data komposit berupa peta yang menampilkan titik sebaran manifestasi panas bumi permukaan, kelurusan struktur patahan (fault/lineament), dan kontur gradien suhu.

## Constraints
- Jangan berasumsi bahwa area tepat di bawah mata air panas adalah pusat reservoir, mengingat fluida panas dapat mengalir jauh secara lateral (outflow).
- Hindari penentuan estimasi kedalaman reservoir hanya bermodalkan peta permukaan tanpa adanya data pengukuran vertikal (seperti Magnetotelluric/MT).
- Argumen interpretasi hanya boleh didasarkan pada zona tumpang tindih (overlay) dari data-data yang tersedia.
- Jangan menyertakan estimasi besaran kapasitas daya energi listrik (dalam Megawatt) pada tahapan analisis kualitatif ini.
- Berikan catatan khusus apabila terdapat data yang saling bertentangan (contoh: terdapat mata air panas bersuhu tinggi namun tidak didukung oleh anomali struktur geologi di sekitarnya).

## Output
Hasil analisis memuat:

- Area dengan konsentrasi manifestasi panas bumi tertinggi.
- Identifikasi awal untuk zona upflow (pusat semburan panas) dan outflow (aliran lateral).
- Struktur geologi yang diduga kuat bertindak sebagai jalur naiknya fluida hidrotermal.
- Pemetaan zonasi wilayah prospek berdasarkan tingkat prioritas (tinggi hingga rendah).
- Rekomendasi titik lokasi atau area fokus untuk pelaksanaan survei Magnetotelluric (MT) selanjutnya.

## Quality Criteria
Hasil evaluasi harus memperlihatkan kemampuan sintesis data spasial yang komprehensif. Dasar penentuan zona prospek harus selalu merujuk pada area persinggungan antara tingginya anomali suhu dan kehadiran struktur patahan yang bertindak sebagai zona permeabilitas tinggi.

## Verification
Pastikan bahwa interpretasi zona upflow memang didukung oleh bukti anomali struktural yang terpusat, bukan sekadar wilayah limpasan fluida (outflow) yang berada di elevasi lebih rendah. Jika kesimpulan prospek melenceng jauh dari struktur pengontrolnya, tinjau ulang alur logika analisis spasialnya.

# 5. ANALISIS MIKROSEISMIK UNTUK KESTABILAN LAHAN
## Role
Kita berperan sebagai mahasiswa geofisika lingkungan dan geoteknik yang sedang mengevaluasi tingkat kerentanan seismik pada suatu tata ruang wilayah.

## Task
Mendeliniasi zona rawan resonansi gempa dan potensi longsor/likuifaksi menggunakan analisis nilai frekuensi dominan (f0) dan faktor amplifikasi (A0) yang diekstrak dari data mikroseismik.

## Context
Metode mikroseismik memanfaatkan getaran alami bumi (ambient noise) untuk mengetahui karakteristik dinamis tanah lokal. Melalui penerapan teknik Horizontal to Vertical Spectral Ratio (HVSR), kita dapat memperoleh nilai frekuensi dominan dan tingkat amplifikasi tanah. Area dengan frekuensi dominan rendah namun nilai amplifikasinya tinggi umumnya merupakan lapisan sedimen tebal dan lunak. Area ini sangat rentan mengalami guncangan keras, resonansi bangunan, atau likuifaksi saat terjadi gempa bumi.

## Input
Data berupa peta sebaran spasial atau tabel kontur yang memuat nilai frekuensi dominan (Hz) dan nilai amplifikasi dari hasil perekaman alat mikroseismik di berbagai titik stasiun pengukuran.

## Constraints
- Jangan menyamakan besaran nilai amplifikasi dari mikroseismik secara mutlak dengan skala kerusakan gempa nyata di lapangan.
- Dilarang mendiagnosis bahaya likuifaksi secara pasti apabila tidak ada informasi tambahan geoteknik mengenai kedalaman muka air tanah dangkal.
- Kesimpulan akhir harus dibatasi pada penentuan tingkat Indeks Kerentanan Seismik (Kg) tanah lokal.
- Lakukan pemisahan analisis antara area batuan dasar (bedrock) yang keras dengan area endapan sedimen lunak secara objektif.
- Apabila kurva HVSR menunjukkan banyak puncak (multi-peak), jelaskan bahwa kondisi pelapisan bawah permukaan di area tersebut sangat kompleks dan tingkat ketidakpastiannya tinggi.

## Output
Hasil analisis memuat:

- Distribusi area yang memiliki nilai frekuensi dominan rendah dan tinggi.
- Pemetaan zona dengan tingkat amplifikasi tanah terkuat.
- Estimasi persentase area sedimen tebal berbanding area singkapan batuan keras.
- Lokalisasi zona yang memiliki potensi nilai Indeks Kerentanan Seismik (Kg) tertinggi.
- Rekomendasi mitigasi praktis yang ditujukan untuk perancangan tata ruang wilayah atau parameter standar konstruksi bangunan sipil.

## Quality Criteria
Laporan analisis harus memadukan parameter fisika gelombang dengan kondisi keteknikan secara presisi. Area dengan frekuensi rendah dan amplifikasi tinggi harus secara tegas diidentifikasi sebagai zona endapan sedimen tebal yang rawan goncangan. Susunan kalimat yang digunakan harus terukur, objektif, dan berorientasi pada mitigasi bencana kebencanaan.

## Verification
Periksa kesesuaian pola distribusi spasial pada peta dengan kondisi morfologi regional. Area yang disimpulkan sebagai lapisan sedimen tebal (frekuensi rendah, amplifikasi tinggi) idealnya berlokasi di dataran rendah, wilayah pesisir, atau bantaran sungai; bukan di area perbukitan batuan masif. Apabila ditemukan anomali yang berlawanan dengan topografi, laporkan hal tersebut sebagai anomali lokal yang membutuhkan penyelidikan geologi teknik lebih mendalam.
